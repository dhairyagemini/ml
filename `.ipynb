```python
import pandas as pd

marine_clean = pd.read_csv('country_marine_protected_area.csv')
terrestrial_clean = pd.read_csv('country_terrestrial_protected_area.csv')
vulnerable_clean = pd.read_csv('country_vunerable.csv')
redlist_clean = pd.read_csv('red_list_index_country_timeseries.csv')

# Combine Marine and Terrestrial datasets
protected_area = pd.merge(
    marine_clean,
    terrestrial_clean,
    on=['COU', 'Country', 'Year'],
    how='outer',
    suffixes=('_Marine', '_Terrestrial')
)

combined_data = pd.merge(
    protected_area,
    redlist_clean,
    on=['COU', 'Country', 'Year'],
    how='outer'
)

# Make Value numeric
vulnerable_clean['Value'] = pd.to_numeric(
    vulnerable_clean['Value'],
    errors='coerce'
)

# Select the main biodiversity indicators
main_indicators = [
    'TOT_KNOWN',
    'VULNERABLE',
    'ENDANGERED',
    'CRITICAL',
    'THREATENED',
    'THREAT_PERCENT'
]

vulnerable_main = vulnerable_clean[
    vulnerable_clean['IUCN'].isin(main_indicators)
].copy()

vulnerability_country = vulnerable_main.pivot_table(
    index=['COU', 'Country'],
    columns='IUCN',
    values='Value',
    aggfunc='sum'
).reset_index()

vulnerability_country.columns.name = None

final_combined = pd.merge(
    combined_data,
    vulnerability_country,
    on=['COU', 'Country'],
    how='left'
)

# ==========================================
# STEP 6: EXPLORATORY DATA ANALYSIS
# ==========================================

print("Dataset Shape:")
print(final_combined.shape)

print("\nColumn Names:")
print(final_combined.columns.tolist())

print("\nStatistical Summary:")
display(final_combined.describe())
```

```python
# ==========================================
# ML PREPARATION
# ==========================================

from sklearn.impute import SimpleImputer
from sklearn.model_selection import train_test_split

# Keep only numerical data
ml_data = final_combined.select_dtypes(include='number').copy()

# Remove rows where the target is unavailable
ml_data = ml_data.dropna(subset=['Red_List_Index'])

# Separate features and target
X = ml_data.drop(columns=['Red_List_Index'])
y = ml_data['Red_List_Index']

# Replace infinite values
X = X.replace([float('inf'), float('-inf')], float('nan'))

# Fill missing feature values using median
imputer = SimpleImputer(strategy='median')
X = imputer.fit_transform(X)

# Train/test split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("ML dataset:", ml_data.shape)
print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
print("Number of features:", X_train.shape[1])
```

```python
# ==========================================
# STEP 9: BASELINE ML MODELS
# ==========================================

from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.svm import SVR

# Create models
models = {
    "Linear Regression": LinearRegression(),
    "Decision Tree": DecisionTreeRegressor(
        random_state=42
    ),
    "Random Forest": RandomForestRegressor(
        n_estimators=100,
        random_state=42,
        n_jobs=-1
    ),
    "SVR": SVR()
}

# Train models
trained_models = {}

for name, model in models.items():
    print("Training:", name)

    model.fit(X_train, y_train)

    trained_models[name] = model

print("\nAll models trained successfully!")
```

```python
# ==========================================
# STEP 9B: MODEL EVALUATION
# ==========================================

from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
import numpy as np
import pandas as pd

results = []

for name, model in trained_models.items():

    predictions = model.predict(X_test)

    mae = mean_absolute_error(
        y_test,
        predictions
    )

    rmse = np.sqrt(
        mean_squared_error(
            y_test,
            predictions
        )
    )

    r2 = r2_score(
        y_test,
        predictions
    )

    results.append({
        "Model": name,
        "MAE": mae,
        "RMSE": rmse,
        "R2": r2
    })

results_df = pd.DataFrame(results)

display(results_df)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

plt.bar(
    results_df["Model"],
    results_df["R2"]
)

plt.xlabel("Model")
plt.ylabel("R² Score")
plt.title("Baseline ML Model Comparison - R² Score")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```
